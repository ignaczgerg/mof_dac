```python
import matplotlib.pyplot as plt
import scienceplots
from mpl_toolkits.mplot3d import Axes3D
from ase.io import read
import numpy as np
import seaborn as sns

plt.style.use("science")
plt.rcParams.update({'font.size': 14})  # Increase font size globally

# Load the CIF file
cif_path = "AKEQUE_full_p8-geo_opt_odac_mepoml.cif"
atoms = read(cif_path)

# Extract atomic positions
positions = atoms.get_positions()
elements = atoms.get_chemical_symbols()

# Assign colors based on element type
unique_elements = list(set(elements))
palette = sns.color_palette("husl", len(unique_elements))
color_map = {el: palette[i] for i, el in enumerate(unique_elements)}
colors = [color_map[el] for el in elements]

# Create a 3D scatter plot
fig = plt.figure(figsize=(8, 5))
ax = fig.add_subplot(111, projection='3d')
sc = ax.scatter(positions[:, 0], positions[:, 1], positions[:, 2], c=colors, s=50, edgecolors='k', alpha=0.8)

# Add legend with correct colors
legend_patches = [plt.Line2D([0], [0], marker='o', color=color_map[el], markerfacecolor=color_map[el], markersize=12, label=el) for el in unique_elements]
ax.legend(handles=legend_patches, title="Elements", loc="upper center", bbox_to_anchor=(0.5, 1.08), ncol=5, fontsize=12, title_fontsize=14)

# Labels and aesthetics
# ax.set_xlabel("X-axis (Å)", fontsize=14)
# ax.set_ylabel("Y-axis (Å)", fontsize=14)
# ax.set_zlabel("", fontsize=14, labelpad=10)
# ax.set_title("3D Point Cloud Visualization of CIF Structure", fontsize=16)
# plt.tight_layout()

plt.show()
```
